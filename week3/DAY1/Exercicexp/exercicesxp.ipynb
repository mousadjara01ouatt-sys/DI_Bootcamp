```python
import numpy as np
```

Exercice 1 : Création et manipulation de réseaux
Créez un tableau NumPy 1D contenant des nombres de 0 à 9.

```python
tableau1 = np.arange(10)
print(tableau1)
```

Exercice 2 : Conversion de type et opérations sur l’array
Convertissez la liste `[3.14, 2.17, 0, 1, 2]` en tableau NumPy et convertissez son type de données en entier.

```python
liste_depart = [3.14, 2.17, 0, 1, 2]
tableau2 = np.array(liste_depart).astype(int)
print(tableau2)
```

Exercice 3 : Travailler avec des réseaux multidimensionnels
Créez un tableau NumPy 3x3 avec des valeurs allant de 1 à 9.

```python
tableau3 = np.arange(1, 10).reshape(3, 3)
print(tableau3)
```

Exercice 4 : Créer un tableau multidimensionnel avec des nombres aléatoires
Créez un tableau NumPy 2D de formes (4, 5) remplies de nombres aléatoires.

```python
tableau4 = np.random.rand(4, 5)
print(tableau4)
```

 Exercise 5 : Indexing Arrays
Sélectionner la deuxième ligne du tableau 2D donné.

```python
matrice = np.array([[21, 22, 23, 22, 22],
                    [20, 21, 22, 23, 24],
                    [21, 22, 23, 22, 22]])

deuxieme_ligne = matrice[1]

print(deuxieme_ligne)
```

Exercice 6 : Inverser les éléments
Inversez l’ordre des éléments dans un tableau NumPy 1D donné.

```python
tableau6 = np.arange(10)
tableau_inverse = tableau6[::-1]
print(tableau_inverse)
```

 Exercice 7 : Matrice d’identité
 Créez une matrice d’identité 4x4 en utilisant NumPy.

```python
matrice_identite = np.eye(4)
print(matrice_identite)
```

Exercice 8 : Fonctions agrégées simples
Trouvez la somme et la moyenne d’un tableau 1D donné (de 0 à 9).

```python
tableau8 = np.arange(10)
somme = tableau8.sum()
moyenne = tableau8.mean()

print(f"Sum: {somme}, Average: {moyenne}")
```

Exercice 9 : Créer un tableau et changer sa structure
Créer un tableau NumPy avec des éléments allant de 1 à 20, puis remodelez-le en une matrice 4x5.

```python
tableau9 = np.arange(1, 21).reshape(4, 5)
print(tableau9)
```

Exercice 10 : Conditional Selection of Values
Extraire tous les nombres impairs d’un tableau NumPy donné (de 1 à 9).

```python
tableau10 = np.arange(1, 10)
nombres_impairs = tableau10[tableau10 % 2 != 0]
print(nombres_impairs)
```
