## Exercice 1 : Définir le problème et collecter les données pour la prédiction du défaut de paiement des prêts

### Énoncé du problème
Créer un modèle de classification binaire pour prédire si un demandeur de prêt va faire défaut de paiement, afin de réduire les pertes financières de la banque tout en validant les profils solvables.

### Données nécessaires
* Données démographiques : Âge, niveau d'éducation, statut d'emploi.
* Données financières : Revenus globaux, taux d'endettement, montant et durée du prêt.
* Historique de crédit : Retards de paiement et score de crédit.

### Sources de données
* Internes : Historique des comptes et des anciennes demandes de l'institution.
* Externes : Bureaux de crédit et agences de notation.

## Exercice 2 : Sélection de fonctionnalités et choix de modèle pour la prédiction du défaut de paiement des prêts

Les variables les plus pertinentes du jeu de données Loan Prediction sont :
* Credit_History : Indique la fiabilité passée, élément le plus prédictif du comportement futur.
* ApplicantIncome et CoapplicantIncome : Déterminent la capacité financière brute à assumer la dette.
* LoanAmount et Loan_Amount_Term : Calculent le poids de la mensualité sur le budget

Justification : L'exclusion des critères démographiques purs (comme le genre) permet d'éliminer les biais discriminatoires et de se concentrer uniquement sur le risque financier objectif.

## Exercice 3 : Entraînement, évaluation et optimisation du modèle

### Choix du modèle
Les modèles basés sur des arbres de décision comme XGBoost ou LightGBM sont idéaux. Ils gèrent efficacement les relations non linéaires, les valeurs manquantes et les données tabulaires asymétriques.

### Métriques d'évaluation
Les cas de défaut étant minoritaires, l'exactitude (Accuracy) est trompeuse. Il faut utiliser :
* Le rappel (Recall) : Métrique prioritaire pour intercepter le maximum de clients non solvables.
* La précision : Pour éviter de rejeter à tort les bons clients.
* Le score F1 et l'AUC-ROC : Pour mesurer l'équilibre global de la classification.

## Exercice 4 : Concevoir des solutions d’apprentissage automatique pour des problèmes spécifiques

### Prédiction des prix des actions
* Type adapté : Apprentissage supervisé (Régression).
* Justification : On se base sur des données passées connues (prix historiques) pour prédire une valeur numérique continue future.

### Organisation d’une bibliothèque de livres
* Type adapté : Apprentissage non supervisé (Clustering).
* Justification : On regroupe les livres par similarités de contenu ou de texte sans utiliser de catégories ou d'étiquettes définies à l'avance.

### Robot dans un labyrinthe
* Type adapté : Apprentissage par renforcement.
* Justification : Le robot doit apprendre la séquence d'actions optimales par lui-même via un système de récompenses et de pénalités basé sur ses essais-erreurs.

## Exercice 5 : Concevoir une stratégie d’évaluation pour différents modèles d’apprentissage automatique

### 1. Modèle supervisé (Classification)
* Stratégie : Validation croisée (K-Fold) pour valider la stabilité et analyse des courbes ROC.
* Métriques : Précision, Rappel, F1-Score.
* Limite : Forte dépendance aux biais historiques présents dans les données d'entraînement.

### 2. Modèle non supervisé (Clustering)
* Stratégie : Méthode du coude (Elbow) pour fixer le nombre idéal de groupes.
* Métriques : Score de Silhouette pour mesurer la cohésion et la séparation des groupes.
* Limite : Absence de vérité absolue, la pertinence finale des groupes dépend d'une interprétation humaine.

### 3. Modèle d'apprentissage par renforcement
* Stratégie : Suivi de l'évolution de la courbe de récompense cumulative et de la convergence de la politique.
* Limite : La définition de la fonction de récompense est complexe ; l'agent peut exploiter des failles sans réellement résoudre le problème initial.
